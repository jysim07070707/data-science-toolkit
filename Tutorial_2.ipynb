```python
import numpy as np

array_2x2 = np.array([
    [1, 2],
    [3, 4]
])

print("This is my numpy 2x2 array")
print(array_2x2)
```

```python
random3x4 = np.random.rand(3,4)

print("Random 3x4 array")
print(random3x4)
```

```python
ones_array = np.ones ((3, 3))
zeros_array = np.zeros ((3, 3))

print ("Array of Ones:")
print (ones_array)

print ("\nArray of Zeros:")
print(zeros_array)
```

```python
random_6x3 = np.random.rand(6, 3)
transposed_3x6 = random_6x3.T
print("Original Array")
print(random_6x3)
print("Transposed Array:")
print(transposed_3x6)
```

```python
matrix_A = np.random.rand(2,3)
matrix_B = np.random.rand(3,4)

print ("Matrix A:")
print(matrix_A)
print ("Matrix B:")
print(matrix_B)

dot_product = np.dot(matrix_A, matrix_B)
print("The dot product of the two matrices is:")
print(dot_product)
```
